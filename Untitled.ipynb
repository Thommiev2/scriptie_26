```python
import pandas as pd
import json
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
output_file_name = "2026-06-04_16-50-03.csv"
df = pd.read_csv(Path('output/summarization scoring/2026-06-04_16-50-03.csv'))
df_ts = pd.read_csv(Path('output/transcription scoring/2026-06-04_16-50-03.csv')).rename(columns={'model': 'asr_model'})

scores = ['bert_precission', 'bert_recall', 'bert_f1', 'rouge', 'compression_rate', 'density', 'coverage', 'summac']
t_scores = ['simdist', 'wer']

df['bertscore'] = df['bertscore'].apply(lambda x: [None, None, None] if pd.isna(x) else x[1:-1].split(', '))
df['bert_precission'] = df['bertscore'].apply(lambda x: None if pd.isna(x[0]) else float(x[0]))
df['bert_recall'] = df['bertscore'].apply(lambda x: None if pd.isna(x[1]) else float(x[1]))
df['bert_f1'] = df['bertscore'].apply(lambda x: None if pd.isna(x[2]) else float(x[2]))
df = df.drop(columns=['bertscore'])

asr_models = df['asr_model'].unique()
genai_models = df['genai_model'].unique()

df.describe()
```

```python
print(asr_models)
print(genai_models)
```

```python
def weighted_average_transcript(transcript_data):
    d = {}
    weights = transcript_data['num_words'].fillna(0)
    total_weight = weights.sum()

    for col in t_scores:
        d[col] = (transcript_data[col] * weights).sum() / total_weight
    return pd.Series(d)
```

```python
# DATAFRAME CONTAINING AVERAGE SCORES FOR EACH MODEL WITH TRANSCRIPT SCORES ADJUSTED FOR LENGTH

asr_m = df.groupby('asr_model')[scores].mean()
asr_mt = df_ts.groupby('asr_model')[t_scores].mean()
asr_model_average_scores = pd.concat([asr_m, asr_mt], axis=1)
asr_model_average_scores
```

```python
# TRANSCRIPT SCORES ADJUSTED FOR GT LENGTH

asr_mt = df_ts.groupby('asr_model')[t_scores].mean()
asr_mt_adj = df_ts.groupby('asr_model').apply(weighted_average_transcript).rename(columns={'wer': 'adj wer', 'simdist': 'adj simdist'})

asr_model_adj_scores = pd.concat([asr_mt, asr_mt_adj], axis=1)
asr_model_adj_scores
```

```python
genai_model_average_scores = df.groupby('genai_model')[scores].mean()
genai_model_average_scores
```

```python
combined_df = pd.merge(
    df, 
    df_ts, 
    on=['name', 'category', 'asr_model'], 
    how='left'
)
combined_df.head()
```

```python
# Set a clean visual style
sns.set_theme(style="whitegrid")

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_ts,        # Replace with your actual dataframe name containing these columns
    x='wer', 
    y='simdist', 
    hue='model',       # Colors the dots by ASR model to spot trends easily
    style='model',     # Uses different marker shapes for better readability
    s=100              # Adjusts the size of the data points
)

# Add titles and clear labels
plt.title('WER vs. Similarity Distance across ASR Models', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Word Error Rate', fontsize=12)
plt.ylabel('Similarity Distance', fontsize=12)

# Move the legend outside the plot area so it doesn't cover your data points
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='ASR Models')

plt.tight_layout()
plt.show()
```

```python
# 1. List out all the numerical score columns you want to correlate
all_scores = ['wer', 'simdist', 'bert_f1', 'rouge', 'summac', 'compression_rate', 'density', 'coverage']

# 3. Plot a heatmap for each GenAI model side-by-side
fig, axes = plt.subplots(1, len(genai_models), figsize=(18, 7))

for i, model in enumerate(genai_models):
    # Filter data for the specific GenAI model
    model_df = combined_df[combined_df['genai_model'] == model]
    
    # Calculate the correlation matrix
    corr_matrix = model_df[all_scores].corr()
    
    # Generate the heatmap
    sns.heatmap(
        corr_matrix, 
        annot=True,             # Shows the correlation numbers in the squares
        cmap='coolwarm',        # Red = positive correlation, Blue = negative correlation
        fmt=".2f",              # Rounds numbers to 2 decimal places
        vmin=-1, vmax=1,        # Sets the scale bounds strictly between -1 and 1
        ax=axes[i],
        cbar=(i == len(genai_models) - 1) # Only show the color bar on the last plot
    )
    axes[i].set_title(f'Correlation Matrix: {model}', fontsize=14, fontweight='bold', pad=10)

plt.tight_layout()
plt.show()
```
